# Theozyme-steer CPU demo on Colab

This notebook runs the **synthetic CPU demo**, not RFdiffusion2. GPU is optional for the runtime inventory only. Upload the delivered ZIP below. Colab execution has not been validated in the build environment.


```python
from google.colab import files
from pathlib import Path
import zipfile
uploaded = files.upload()
archive = next(name for name in uploaded if name.endswith('.zip'))
root = Path('/content/theozyme_project').resolve()
root.mkdir(exist_ok=True)
with zipfile.ZipFile(archive) as z:
    for item in z.infolist():
        target = (root / item.filename).resolve()
        if not target.is_relative_to(root):
            raise ValueError('Unsafe archive path')
    z.extractall(root)
project = root / 'theozyme-steer'
%cd $project
```
